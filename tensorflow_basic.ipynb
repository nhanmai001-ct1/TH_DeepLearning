BÀI TẬP 5: Cho tensor X = [[1, 2], [3, 4], [5, 6]]. Hãy shuffle X theo chiều thứ nhất (sử dụng

```python
import tensorflow as tf

X = tf.constant([
    [1, 2],
    [3, 4],
    [5, 6]
])

# shuffle theo chiều thứ nhất (shuffle các dòng)
X_shuffle = tf.random.shuffle(X)

print(X_shuffle)
```

BÀI 6: Tạo ngẫu nhiên 2 tensor kích thước x = (3, 2) và giá trị trong khoảng [5, 10], y = (2, 3) và giá
trị [15, 20] tương ứng. Hãy giữ giá trị đường chéo của x và thực hiện x + y.

```python
import tensorflow as tf

# =========================
# Tạo tensor x kích thước (3,2)
# giá trị trong khoảng [5,10]
# =========================

x = tf.random.uniform(
    shape=(3, 2),
    minval=5,
    maxval=10,
    dtype=tf.int32
)

# =========================
# Tạo tensor y kích thước (2,3)
# giá trị trong khoảng [15,20]
# =========================

y = tf.random.uniform(
    shape=(2, 3),
    minval=15,
    maxval=20,
    dtype=tf.int32
)

print("Tensor x:\n", x)
print("Tensor y:\n", y)

# =========================
# Giữ giá trị đường chéo của x
# =========================

diag_x = tf.linalg.diag_part(x)

print("Đường chéo của x:\n", diag_x)

# =========================
# x + y
# =========================
# Vì x(3,2) và y(2,3) khác shape
# nên cần transpose y -> (3,2)
# =========================

y_t = tf.transpose(y)

result = x + y_t

print("x + transpose(y):\n", result)
```
